# Gaussian Quadrature

Use weighted sample points to integrate a polynomial or smooth function accurately.

```python
import numpy as np

# Two-point Gauss-Legendre rule on [-1, 1]
nodes = np.array([-1/np.sqrt(3), 1/np.sqrt(3)])
weights = np.array([1., 1.])

def gauss_two_point(f, a, b):
    midpoint = (a+b)/2
    half = (b-a)/2
    return half * sum(w * f(midpoint + half*t) for t, w in zip(nodes, weights))

print('integral x^4 on [0, 1]:', gauss_two_point(lambda x: x**4, 0, 1))
print('exact:', 0.2)
```

Gaussian quadrature can achieve high accuracy with few evaluations when the nodes and weights are chosen well.
