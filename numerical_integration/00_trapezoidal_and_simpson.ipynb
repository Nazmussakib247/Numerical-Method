# Trapezoidal and Simpson Rules

Approximate a definite integral using composite Newton-Cotes formulas.

```python
import math

def trapezoidal(f, a, b, n):
    h = (b-a)/n
    return h * (0.5*f(a) + sum(f(a+i*h) for i in range(1,n)) + 0.5*f(b))

def simpson(f, a, b, n):
    if n % 2: raise ValueError('n must be even for Simpson rule')
    h = (b-a)/n
    odd = sum(f(a+i*h) for i in range(1,n,2))
    even = sum(f(a+i*h) for i in range(2,n,2))
    return h/3 * (f(a)+f(b)+4*odd+2*even)

f = math.sin
print('trapezoidal:', trapezoidal(f, 0, math.pi, 100))
print('Simpson:', simpson(f, 0, math.pi, 100))
print('exact:', 2.0)
```

Increasing the number of subintervals generally reduces error for smooth functions.
