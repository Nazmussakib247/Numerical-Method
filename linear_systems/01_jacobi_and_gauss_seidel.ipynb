# Jacobi and Gauss-Seidel Iteration

Compare two iterative methods for diagonally dominant linear systems.

```python
import numpy as np

A = np.array([[10., 1., 1.], [2., 10., 1.], [2., 2., 10.]])
b = np.array([12., 13., 14.])

def jacobi(A, b, iterations=50):
    x = np.zeros_like(b)
    D = np.diag(A)
    R = A - np.diagflat(D)
    for _ in range(iterations):
        x = (b - R @ x) / D
    return x

def gauss_seidel(A, b, iterations=50):
    x = np.zeros_like(b)
    for _ in range(iterations):
        for i in range(len(b)):
            x[i] = (b[i] - A[i, :i] @ x[:i] - A[i, i+1:] @ x[i+1:]) / A[i, i]
    return x

print('Jacobi:', np.round(jacobi(A, b), 6))
print('Gauss-Seidel:', np.round(gauss_seidel(A, b), 6))
print('Direct:', np.round(np.linalg.solve(A, b), 6))
```

Gauss-Seidel reuses fresh values immediately and commonly converges faster on suitable systems.
