# Gaussian Elimination

Solve a small linear system by forward elimination followed by back substitution.

```python
import numpy as np

def gaussian_elimination(A, b):
    A = np.asarray(A, dtype=float).copy()
    b = np.asarray(b, dtype=float).copy()
    n = len(b)
    for pivot in range(n):
        row = pivot + np.argmax(np.abs(A[pivot:, pivot]))
        if abs(A[row, pivot]) < 1e-12:
            raise ValueError('Singular system.')
        A[[pivot, row]] = A[[row, pivot]]
        b[[pivot, row]] = b[[row, pivot]]
        for r in range(pivot + 1, n):
            factor = A[r, pivot] / A[pivot, pivot]
            A[r, pivot:] -= factor * A[pivot, pivot:]
            b[r] -= factor * b[pivot]
    x = np.zeros(n)
    for r in range(n - 1, -1, -1):
        x[r] = (b[r] - A[r, r+1:] @ x[r+1:]) / A[r, r]
    return x

A = [[3, 2, -1], [2, -2, 4], [-1, 0.5, -1]]
b = [1, -2, 0]
x = gaussian_elimination(A, b)
print('solution:', np.round(x, 6))
print('residual norm:', np.linalg.norm(np.asarray(A) @ x - b))
```

Partial pivoting improves numerical stability by selecting a large pivot in each column.
