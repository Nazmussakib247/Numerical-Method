# LU Factorization

Factor a matrix into lower and upper triangular matrices and solve through two triangular systems.

```python
import numpy as np

A = np.array([[4., 3.], [6., 3.]])
L = np.eye(2)
U = A.copy()
for i in range(1, 2):
    L[i, 0] = U[i, 0] / U[0, 0]
    U[i] -= L[i, 0] * U[0]
print('L =\n', L)
print('U =\n', U)
b = np.array([10., 12.])
y = np.linalg.solve(L, b)
x = np.linalg.solve(U, y)
print('solution:', x)
print('reconstruction error:', np.linalg.norm(L @ U - A))
```

LU is useful when the same coefficient matrix is solved for multiple right-hand sides.
