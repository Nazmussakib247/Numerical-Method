# Finite Difference Derivatives

Estimate first and second derivatives with forward, backward, and central differences.

```python
import math

f = math.sin
x = 1.0
h = 1e-4
forward = (f(x+h)-f(x))/h
backward = (f(x)-f(x-h))/h
central = (f(x+h)-f(x-h))/(2*h)
second = (f(x+h)-2*f(x)+f(x-h))/h**2
print('forward:', forward)
print('backward:', backward)
print('central:', central)
print('second derivative:', second)
print('exact first:', math.cos(x), 'exact second:', -math.sin(x))
```

Central differences often balance truncation error more effectively than one-sided formulas.
