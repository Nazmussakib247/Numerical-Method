# Richardson Extrapolation

Combine derivative estimates at two step sizes to cancel leading-order error.

```python
import math

def central(f, x, h):
    return (f(x+h)-f(x-h))/(2*h)

def richardson(f, x, h):
    d_h = central(f, x, h)
    d_half = central(f, x, h/2)
    return d_half + (d_half-d_h)/3

x, h = 1.0, 0.1
print('central:', central(math.sin, x, h))
print('Richardson:', richardson(math.sin, x, h))
print('exact:', math.cos(x))
```

Extrapolation improves an approximation when the error follows a predictable power of the step size.
