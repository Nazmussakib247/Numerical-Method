# Fourth-Order Runge-Kutta

Solve an initial-value problem using the classical RK4 method.

```python
import math

def rk4(f, t0, y0, h, steps):
    t, y = t0, y0
    for _ in range(steps):
        k1 = f(t, y)
        k2 = f(t+h/2, y+h*k1/2)
        k3 = f(t+h/2, y+h*k2/2)
        k4 = f(t+h, y+h*k3)
        y += h*(k1+2*k2+2*k3+k4)/6
        t += h
    return t, y

t, y = rk4(lambda t, y: y, 0, 1, 0.1, 10)
print('RK4 y(1):', y)
print('Exact y(1):', math.e)
```

RK4 uses four slope estimates per step and is much more accurate for smooth problems.
