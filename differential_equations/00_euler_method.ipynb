# Euler Method for an ODE

Approximate the solution of a first-order initial-value problem with explicit Euler steps.

```python
import math

def euler(f, t0, y0, h, steps):
    t, y = t0, y0
    points = [(t, y)]
    for _ in range(steps):
        y += h * f(t, y)
        t += h
        points.append((t, y))
    return points

points = euler(lambda t, y: y, 0, 1, 0.1, 10)
print('Euler y(1):', points[-1][1])
print('Exact y(1):', math.e)
```

Euler is simple and inexpensive, but its global error is proportional to the step size.
