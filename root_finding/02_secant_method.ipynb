# Secant Method

Approximate derivatives using two previous points and solve a nonlinear equation without an explicit derivative.

```python
def secant(f, x0, x1, tol=1e-9, max_iter=100):
    history = [x0, x1]
    for _ in range(max_iter):
        denominator = f(x1) - f(x0)
        if denominator == 0:
            raise ZeroDivisionError('Secant slope is zero.')
        x2 = x1 - f(x1) * (x1 - x0) / denominator
        history.append(x2)
        if abs(x2 - x1) < tol:
            return x2, history
        x0, x1 = x1, x2
    return x1, history

root, history = secant(lambda x: x**3 - x - 2, 1, 2)
print(f'root = {root:.10f}, iterations = {len(history)-2}')
```

The secant method often converges faster than bisection while avoiding an analytic derivative.
