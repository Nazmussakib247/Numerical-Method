# Bisection Method

Find a root of a continuous function by repeatedly halving a sign-changing interval.

We solve $f(x)=x^3-x-2$ and maintain the invariant $f(a)f(b)<0$.

```python
def bisection(f, a, b, tol=1e-8, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError('The interval must bracket a root.')
    history = []
    for _ in range(max_iter):
        c = (a + b) / 2
        history.append((a, b, c, f(c)))
        if abs(f(c)) < tol or (b - a) / 2 < tol:
            return c, history
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2, history

root, history = bisection(lambda x: x**3 - x - 2, 1, 2)
print(f'root = {root:.8f}, iterations = {len(history)}')
```

The method is reliable when continuity and a valid bracket are available, although convergence is linear.
