# Newton-Raphson Method

Use a tangent-line update to obtain fast local convergence for a nonlinear equation.

Newton’s update is $x_{k+1}=x_k-f(x_k)/f'(x_k)$.

```python
def newton(f, df, x0, tol=1e-10, max_iter=50):
    x = float(x0)
    history = [x]
    for _ in range(max_iter):
        step = f(x) / df(x)
        x -= step
        history.append(x)
        if abs(step) < tol:
            break
    return x, history

f = lambda x: x**3 - 2*x - 5
df = lambda x: 3*x**2 - 2
root, history = newton(f, df, 2.0)
print(f'root = {root:.10f}, residual = {abs(f(root)):.2e}')
```

A good initial guess matters because Newton-Raphson can diverge or converge to another root.
