# Newton Forward Interpolation

Use a forward-difference table for equally spaced data points.

```python
import math

def forward_table(values):
    table = [list(values)]
    while len(table[-1]) > 1:
        prev = table[-1]
        table.append([prev[i+1] - prev[i] for i in range(len(prev)-1)])
    return table

def newton_forward(x0, h, values, x):
    table = forward_table(values)
    u = (x - x0) / h
    result = table[0][0]
    term = 1.0
    for order in range(1, len(table)):
        term *= (u - order + 1) / order
        result += term * table[order][0]
    return result

values = [1, 4, 9, 16]
print('f(1.5) ≈', newton_forward(0, 1, values, 1.5))
```

Forward interpolation is convenient near the beginning of a uniformly spaced table.
