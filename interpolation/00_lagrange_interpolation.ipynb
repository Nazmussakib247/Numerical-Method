# Lagrange Interpolation

Construct a polynomial that passes exactly through a set of known data points.

```python
def lagrange(x_data, y_data, x):
    total = 0.0
    for i, (xi, yi) in enumerate(zip(x_data, y_data)):
        basis = 1.0
        for j, xj in enumerate(x_data):
            if i != j:
                basis *= (x - xj) / (xi - xj)
        total += yi * basis
    return total

x_data = [0, 1, 2]
y_data = [1, 3, 2]
for x in [0.5, 1.5]:
    print(x, lagrange(x_data, y_data, x))
```

The basis polynomials make the interpolation property explicit, but high-degree interpolation can be unstable.
