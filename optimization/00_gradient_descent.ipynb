# Gradient Descent

Minimize a differentiable objective by moving repeatedly in the negative gradient direction.

```python
def gradient_descent(gradient, x0, learning_rate=0.1, steps=30):
    x = float(x0)
    history = [x]
    for _ in range(steps):
        x -= learning_rate * gradient(x)
        history.append(x)
    return x, history

minimum, history = gradient_descent(lambda x: 2*(x-3), -5)
print('minimum estimate:', minimum)
print('objective:', (minimum-3)**2)
```

The learning rate controls stability: a value that is too large can overshoot the minimum.
