# Least Squares Normal Equation

Fit a linear model by minimizing the sum of squared residuals.

```python
import numpy as np

X = np.array([[1, 0], [1, 1], [1, 2], [1, 3]], dtype=float)
y = np.array([1.1, 2.9, 5.2, 6.8])
beta = np.linalg.solve(X.T @ X, X.T @ y)
predictions = X @ beta
print('intercept, slope:', np.round(beta, 4))
print('predictions:', np.round(predictions, 3))
print('SSE:', np.sum((y-predictions)**2))
```

The normal equation gives the least-squares solution when the design matrix has full column rank.
